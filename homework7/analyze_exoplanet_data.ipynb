```python
# There are 7 different discover methods
# Transit: This is where a planet is discovered by analysing the brightness of a star and consistantly seeing a dip in brightness. This means an exoplanet is moving across it.
# Radial Velcoty: This is where you analyze the shift in spectral lines which can be caused by the orbiting planet
# Microlensing: This is when a planet in front of a star system passes in front of a background star which causes a magnification of the background star’s light
# Imaging: This is where you capture the actual light coming from a planet. 
# Tranasit Timing Variations: This is where you look at small shifts in already known exoplanet transits likely due to the gravitational pull of another planet
# Ecplis Timing Variations: This is where you analyze ecplising binaries and if there is any variation in timing that can be caused by an orbiting planet
# Disk Kinematics: This is where you look for disturbances in the gas flow of planet's disks
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=small_table, x="pl_bmassj", y="pl_eqt", hue="discoverymethod")
plt.xlabel("Planet mass in jupiter masses")
plt.ylabel("Equilibrium Temperature in k")
plt.title("Mass vs Equilibrium Temperature")
plt.legend(title="Discovery Method", loc='upper left')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=small_table, x=np.log(small_table["pl_bmassj"]), y=np.log(small_table["pl_eqt"]), hue="discoverymethod")
plt.xlabel("Planet mass in jupiter masses")
plt.ylabel("Equilibrium Temperature in k")
plt.title("Mass vs Equilibrium Temperature")
plt.legend(title="Discovery Method", loc='upper left')
plt.show()
```
