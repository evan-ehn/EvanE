```python
median_lat = np.median(df["latitude"])
median_lon = np.median(df["longitude"])
print(median_lat, median_lon)
```

```python
df["Absolute difference from Latitude"] = abs(df["latitude"] - median_lat)
df["Absolute difference from Longitude"] = abs(df["longitude"] - median_lon)
df
```

```python
df.shape
```

```python
# There are 52 and now 6 columns
```

```python
closest_to_med_lat = df["Absolute difference from Latitude"].idxmin()
closest_to_med_lat_state = df.loc[closest_to_med_lat, "name"]
closest_to_med_lat_value = df.loc[closest_to_med_lat, "Absolute difference from Latitude"]

furthest_from_med_lat = df["Absolute difference from Latitude"].idxmax()
furthest_from_med_lat_state = df.loc[furthest_from_med_lat, "name"]
furthest_from_med_lat_value = df.loc[furthest_from_med_lat, "Absolute difference from Latitude"]

closest_to_med_lon = df["Absolute difference from Longitude"].idxmin()
closest_to_med_lon_state = df.loc[closest_to_med_lon, "name"]
closest_to_med_lon_value = df.loc[closest_to_med_lon, "Absolute difference from Longitude"]

furthest_from_med_lon = df["Absolute difference from Longitude"].idxmax()
furthest_from_med_lon_state = df.loc[furthest_from_med_lon, "name"]
furthest_from_med_lon_value = df.loc[furthest_from_med_lon, "Absolute difference from Longitude"]

print(f"state index: {closest_to_med_lat}, state name: {closest_to_med_lat_state}, closest to median latitude value: {closest_to_med_lat_value:.2f}")
print(f"state index: {furthest_from_med_lat}, state name: {furthest_from_med_lat_state}, furthest from median latitude value: {furthest_from_med_lat_value:.2f}")
print(f"state index: {closest_to_med_lon}, state name: {closest_to_med_lon_state}, closest to median longitude value: {closest_to_med_lon_value:.2f}")
print(f"state index: {furthest_from_med_lon}, state name: {furthest_from_med_lon_state}, furthest from median latitude value: {furthest_from_med_lon_value:.2f}")
```

```python
# The year represents the year the of the flight. The month represents the month the flight took off in. The passenger value is the number of passengers on the flight.
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 10))
sns.scatterplot(data=data_set, x='year', y='passengers', ax=axes[0, 0], color = 'green', hue='year')
axes[0, 0].set_title("Year vs Passengers")
sns.scatterplot(data=data_set, x='passengers', y='year', ax=axes[0, 1], color = 'yellow')
axes[0, 1].set_title("Passengers vs Year (Switched Axes)")
sns.histplot(data=data_set, x='passengers', ax=axes[1, 0], color = 'purple')
axes[1, 0].set_title("Passenger Histogram")
sns.boxplot(data=data_set, x='month', y='passengers', ax=axes[1, 1], color = 'red')
axes[1, 1].set_title("Passengers by Month")
plt.show()
```
