```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# sin function
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)
```

```python
plt.figure(figsize = (6, 4))
plt.plot(x, y)
plt.show()
```

```python
y1 = np.sin(x)
y2 = np.cos(x)
plt.figure()
plt.scatter(x, y1, label = "sin(x)", color = "green")
plt.plot(x, y2, label = "cos(x)", color = "orange", linestyle = "-.", linewidth = 8, alpha = 0.3, marker = "*")
plt.xlabel("x value, in radians")
plt.ylabel("f(x)")
plt.legend(loc = "upper right")
plt.show()
```

```python
# subplots:
```

```python
fig, axes = plt.subplots(1, 2, figsize = (10, 4))
axes[0].plot(x, y1, color = "teal", ls = ":")
axes[0].set_title("sine wave")
axes[1].plot(x, y1, color = "purple", ls = "--")
axes[1].set_title("cos wave")
plt.show()
```

```python
fig, axes = plt.subplots(2, 1, figsize = (4, 10))
axes[0].scatter(x, y1, color = "teal", ls = ":", s = 1)
axes[1].scatter(x, y2, color = "red", ls = "-.", s = 10)
axes[0].set_title("sin wave")
axes[1].set_title("cos wave")
plt.show()
```
