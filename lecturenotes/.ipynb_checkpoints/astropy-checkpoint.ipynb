```python
def ideal_gas_lae(number_of_moles, volume, tempurature):
    V = volume *(u.m**3)
    n = number_of_moles*(u.mol)
    R = const.R
    T = tempuratre*u.k
    P = ((n*R*T)/V)*(u.Pa)
    return(P)
```
