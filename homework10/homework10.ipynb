```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from astropy import units as u
from astropy import constants as const
```

```python
data = pd.read_csv("solar_system.csv")
df = pd.DataFrame(data)
print(df.shape)
```

```python
df = df.set_index("Attribute").T
df.index.name = "Planet"
df.reset_index(inplace=True)
df.columns.name = None
```

```python
print(df.shape)
```

```python
# There are 20 columns and 11 rows before and after transposing
# df = df.set_index("Attribute").T: This sets the attributes column containing variables like mass and number of moons and turns it into columns. 
# df.index.name = "Planet": This gives the row indexes for planets a header called planets
# df.reset_index(inplace=True): This converts the rows back into column titles
# df.columns.name = None: This cleans the headers so they look blank. 
# because df.reset_index(inplace=True) pulls the planet names out of the index and turns them into a new column on the left, it adds an extra column.
# This shifts the final shape to 10 rows of planets + 1 header row, and 11 attribute columns + 1 new "Planet" column, resulting in a different dimension structure.
print(df.columns)
```

```python
# Units: Mass (10^24kg), Diameter (km), Density (kg/m^3)Gravity (m/s^2), Escape Velocity (km/s), Rotation Period (hours), Length of Day (hours), Distance from Sun ($10^6$ km), Perihelion ($10^6$ km)
# Aphelion ($10^6$ km), Orbital Period (days), Orbital Velocity (km/s), Orbital Inclination (deg), Obliquity to Orbit (deg), Mean Temperature (C), Surface Pressure (bars)
# No units: Planet, number, and ring system
df
```

```python
# They are all strings. Storing a numeric value as a string rather than a float or int, means you can't manipulate it or use it in any equations.
```

```python
for col in df.columns:
    if col not in ["Planet", "Ring System?", "Global Magnetic Field?"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.dtypes
```

```python
# The 3 types are strings, floats, and integers
# String is just a string of letters with no numerical value. Int is a number that can't include decimals. Float is a number that can include deciamals. 
# Planet, ring system, and magnetic field are all objects because they were excluded int he conversion. They also don't have numerical values. 
```

```python
def attach_units(col):
    new_values = [val * 1e24 * u.kg for val in df[col]]
    df[col] = new_values
    df.rename(columns={col: "Mass (kg)"}, inplace=True)
```

```python
attach_units("Mass (10^24kg)")
print(df["Mass (kg)"])
df
```

```python
sma = (df["Perihelion (10^6 km)"] + df["Aphelion (10^6 km)"]) / 2
semi_major_axis = sma.values * (1e6 * u.km)
aphelion_idx = df.columns.get_loc("Aphelion (10^6 km)")
df.insert(aphelion_idx + 1, "Semi-Major Axis", semi_major_axis)
print("Unit confirmation:", semi_major_axis.unit)
print(df["Semi-Major Axis"].values)
```

```python
orbital_period_years = (df["Orbital Period (days)"].values * u.day).to(u.yr)
df["Orbital Period (years)"] = orbital_period_years
df.drop(columns=["Orbital Period (days)"], inplace=True)
print(df.columns)
print(df["Orbital Period (years)"].values)
fav_planet = "Earth"
fav_idx = df.index[df["Planet"] == fav_planet][0]
fav_period = orbital_period_years[fav_idx]
print(f"{fav_planet}'s orbital period: {fav_period.value:.4f} {fav_period.unit}")
```

```python
# AU is the average distance from Earth to the Sun
print(const.au)
distance = 1*u.au
distance = distance.to(u.km)
print(distance)
```

```python
distance_cols = ["Diameter (km)", "Distance from Sun (10^6 km)", "Perihelion (10^6 km)", "Aphelion (10^6 km)", "Semi-Major Axis"]
fav_planet = "Earth"
fav_idx = df.index[df["Planet"] == fav_planet][0]
for col in distance_cols:
    if col in df.columns:
        if "10^6" in col:
            converted_vals = (df[col].values * 1e6 * u.km).to(u.au)
            new_col_name = col.replace("10^6 km", "AU")
        elif col == "Semi-Major Axis":
            converted_vals = df[col].values.to(u.au)
            new_col_name = "Semi-Major Axis (AU)"
        else:
            converted_vals = (df[col].values * u.km).to(u.au)
            new_col_name = col.replace("km", "AU")
        df[new_col_name] = converted_vals
        df.drop(columns=[col], inplace=True)
        val = df.loc[fav_idx, new_col_name]
        print(f"{fav_planet}'s {new_col_name}: {val.value:.4f} {val.unit}")
```

```python
data_dict = {}
for col in df.columns:
    series = df[col]
    try:
        data_dict[col] = series.values.value
    except AttributeError:
        try:
            data_dict[col] = np.array([x.value if hasattr(x, 'value') else x for x in series])
        except Exception:
            data_dict[col] = series.values
df_clean = pd.DataFrame(data_dict)
df_clean.to_csv("units.csv", index=False) # The line that we were supposed to use didn't work, so I had to find a more convoluted way to save the data file
```
