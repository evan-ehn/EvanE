```python
import numpy as np
```

```python
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)
print(arr*4)
```

```python
print(np.mean(arr))
```

```python
median_arr = np.median(arr)
print(median_arr)
```

```python
print(np.std(arr))
```

```python
print(np.argmax(arr))
print(np.argmin(arr))
```

```python
print(arr[np.argmax(arr)])
print(arr[np.argmin(arr)])
```

```python
print(np.min(arr))
print(np.max(arr))
```

```python
# Create a new array using boolean indexing
new_arr = arr[arr > 20]
print(new_arr)
```
