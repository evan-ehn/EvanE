```python
arr = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])
def prime(n):
    if n < 2:
        return False
    for j in range(2, int(np.sqrt(n)) + 1):
        if n % j == 0:
            return False
    return True

def rows_with_prime(matrix):
    valid_rows = []
    for row in matrix:
        if any(prime(val) for val in row):
            valid_rows.append(row)
    return np.array(valid_rows)

print(rows_with_prime(arr))
```

```python
checkers = np.zeros((8, 8), dtype = int)
```

```python
for i in range(8):
    for j in range(8):
        if (i + j) % 2 != 0:
            checkers[i, j - 1] = 1
print(checkers)
```

```python
for i in range(8):
    for j in range(8):
        if (i + j) % 2 != 0:
            checkers[i, j] = 1
print(checkers)
```

```python
universe = np.array(['galaxy', 'clusters'])
def space(letters, num):
    expanded_universe = []
    for i in letters:
        new_letters = (" " * num).join(list(i))
        expanded_universe.append(new_letters)
    return np.array(expanded_universe)
print(space(universe, 2))
```

```python
np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
def secondDimmest(objects):
    sorted_stars = np.sort(objects, axis=1)
    return sorted_stars[:, -2]
secondDimmest(stars)
```
