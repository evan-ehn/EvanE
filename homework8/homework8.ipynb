```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2*np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
def make_sine_wave(x, A, w):
    y = A*np.sin(w*x)
    return(y)
fig, axes = plt.subplots(1, 3, figsize=(25, 5))
for amp in amplitude:
    axes[0].plot(x, make_sine_wave(x, amp, 1))
    axes[0].set_xlabel("x values")
    axes[0].set_ylabel("amplitude")
    axes[0].set_title("graph of sin curve with varying amplitude")
for freq in frequency:
    axes[1].plot(x, make_sine_wave(x, 1, freq))
    axes[1].set_xlabel("x values")
    axes[1].set_ylabel("ampltude")
    axes[1].set_title("graph of sin curve with varying frequency")
for amp, freq in zip(amplitude, frequency):
    axes[2].plot(x, make_sine_wave(x, amp, freq))
    axes[2].set_xlabel("x values")
    axes[2].set_ylabel("ampltude")
    axes[2].set_title("graph of sin curve with varying frequency and amplitude")

plt.show()
```
