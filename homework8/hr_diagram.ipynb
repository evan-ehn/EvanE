```python
magnitude = data[:, 0]
temperature = data[:, 1]
print(np.min(magnitude), np.argmin(magnitude), np.max(magnitude), np.argmax(magnitude))
print(np.min(temperature), np.argmin(temperature), np.max(temperature), np.argmax(temperature))
```

```python
# The brightest star is star number 1988. The dimmest star is star number 263
```

```python
log_temperature = np.log(temperature)
log_magnitude = np.log(magnitude)
bright_idx = np.argmin(magnitude)
dim_idx = np.argmax(magnitude)
hot_idx = np.argmax(temperature)
cold_idx = np.argmin(temperature)
plt.figure(figsize=(8, 6))
plt.scatter(log_temperature, log_magnitude, label="All Stars")
plt.scatter(log_temperature[bright_idx], log_magnitude[bright_idx], color="black", marker="*", s=300, label="Brightest")
plt.scatter(log_temperature[dim_idx], log_magnitude[dim_idx], color="purple", marker="*", s=300, label="Dimmest")
plt.scatter(log_temperature[hot_idx], log_magnitude[hot_idx], color="green", marker="*", s=300, label="Hottest")
plt.scatter(log_temperature[cold_idx], log_magnitude[cold_idx], color="orange", marker="*", s=300, label="Coldest")# I used s to increase the size of the stars. 
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.legend()
plt.show()
```
